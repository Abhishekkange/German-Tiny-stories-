```python
import json

records = []
with open("/Users/abhishek/German-Tiny-stories-/dataset generation/output (3).jsonl", "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if line:
            records.append(json.loads(line))
```

#### checking if there is duplicate data in dataset

```python
import hashlib, json

def hash_line(obj):
    return hashlib.md5(json.dumps(obj, sort_keys=True).encode()).hexdigest()



seen = {}
count = 0
total = 0
for i, record in enumerate(records):
    total= total+1
    h = hash_line(record)
    
    if h in seen:
        print(f"Row {i} is a duplicate of row {seen[h]}")
        count=count+1
    else:
        seen[h] = i
        
print(count)
print("Total",total)
```
